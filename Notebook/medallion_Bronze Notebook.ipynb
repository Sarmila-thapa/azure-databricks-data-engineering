```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F


# Read file to df
df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/sales_new_catalog/default/landing/customers.csv")
)
```

```python
df.show()
```

```python
bronze_df = df.withColumn("_ingest_ts",     F.current_timestamp())
bronze_df.show()
```

```python
bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "sales_new_catalog.Bronze.customers",
        path="abfss://workshop@sarmila01.dfs.core.windows.net/Bronze/customers"    
    )
```

```python
bronze_df.write \
    .format("csv") \
    .option("header", "true") \
    .mode("overwrite") \
    .saveAsTable(
        "sales_new_catalog.Bronze.customers_csv",
        path="abfss://workshop@sarmila01.dfs.core.windows.net/Bronze/customers_csv"    
    )
```
