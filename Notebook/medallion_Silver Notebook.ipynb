```python
df = spark.sql("select * from sales_new_catalog.Bronze.customers")
df.show()
```

```python
silver_df= df.dropDuplicates(["customer_id"])
silver_df.show()
```

```python
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "sales_new_catalog.Silver.customers",
        path="abfss://workshop@sarmila01.dfs.core.windows.net/Silver/customers"
    )
```

```python
silver_df.write \
    .format("csv") \
    .option("header", "true") \
    .mode("overwrite") \
    .saveAsTable(
        "sales_new_catalog.Silver.customers_csv",
        path="abfss://workshop@sarmila01.dfs.core.windows.net/Silver/customers_csv"    
    )
```
