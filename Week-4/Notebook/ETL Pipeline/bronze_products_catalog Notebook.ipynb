```python
from pyspark.sql import functions as F

source_path = "/Volumes/products_catalog/default/product_pipeline_volume/raw/products/"

df_raw = (
    spark.read
    .option("multiLine", "true")
    .json(source_path)
)

df_bronze = df_raw.select(
    F.explode("products").alias("product")
)

display(df_bronze)
```

```python
print(
    spark.table("products_catalog.default.bronze_products").count()
)
```
