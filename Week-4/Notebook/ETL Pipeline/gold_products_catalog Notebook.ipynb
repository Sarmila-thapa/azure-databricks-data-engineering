```python
from pyspark.sql import functions as F

silver_products = spark.table(
    "products_catalog.default.silver_products"
)
```

```python
from pyspark.sql import functions as F
gold_products = (
    silver_products

    # Final price after discount
    .withColumn(
        "final_price",
        F.round(
            F.col("price") *
            (
                1 -
                F.col("discount_percentage") / 100
            ),
            2
        )
    )

    # Discount amount
    .withColumn(
        "discount_amount",
        F.round(
            F.col("price") *
            F.col("discount_percentage") / 100,
            2
        )
    )

    # Rating category
    .withColumn(
        "rating_category",
        F.when(
            F.col("rating") >= 4.5,
            "Excellent"
        )
        .when(
            F.col("rating") >= 4.0,
            "Good"
        )
        .when(
            F.col("rating") >= 3.0,
            "Average"
        )
        .otherwise("Poor")
    )

    # Stock status
    .withColumn(
        "stock_status",
        F.when(
            F.col("stock") == 0,
            "Out of Stock"
        )
        .when(
            F.col("stock") <= 10,
            "Low Stock"
        )
        .when(
            F.col("stock") <= 50,
            "Medium Stock"
        )
        .otherwise("High Stock")
    )

    # Inventory value
    .withColumn(
        "inventory_value",
        F.round(
            F.col("price") *
            F.col("stock"),
            2
        )
    )

    # Processing timestamp
    .withColumn(
        "gold_processed_at",
        F.current_timestamp()
    )
)
display(gold_products)
```
