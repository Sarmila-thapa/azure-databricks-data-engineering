```python
from pyspark.sql import functions as F
bronze_products = spark.table(
    "products_catalog.default.bronze_products"
)
```

```python
from pyspark.sql import functions as F
silver_products = (
    bronze_products
    .select(
        F.col("product.id").cast("int").alias("product_id"),

        F.trim(F.col("product.title")).alias("title"),

        F.trim(F.col("product.description")).alias("description"),

        F.trim(F.col("product.category")).alias("category"),

        F.trim(F.col("product.brand")).alias("brand"),

        F.trim(F.col("product.sku")).alias("sku"),

        F.col("product.price")
            .cast("decimal(12,2)")
            .alias("price"),

        F.col("product.discountPercentage")
            .cast("decimal(5,2)")
            .alias("discount_percentage"),

        F.col("product.rating")
            .cast("decimal(3,2)")
            .alias("rating"),

        F.col("product.stock")
            .cast("int")
            .alias("stock"),

        F.col("product.weight")
            .cast("decimal(10,2)")
            .alias("weight"),

        F.col("product.dimensions.width")
            .cast("decimal(10,2)")
            .alias("width"),

        F.col("product.dimensions.height")
            .cast("decimal(10,2)")
            .alias("height"),

        F.col("product.dimensions.depth")
            .cast("decimal(10,2)")
            .alias("depth"),

        F.trim(
            F.col("product.warrantyInformation")
        ).alias("warranty_information"),

        F.trim(
            F.col("product.shippingInformation")
        ).alias("shipping_information"),

        F.trim(
            F.col("product.availabilityStatus")
        ).alias("availability_status"),

        F.trim(
            F.col("product.returnPolicy")
        ).alias("return_policy"),

        F.col("product.minimumOrderQuantity")
            .cast("int")
            .alias("minimum_order_quantity"),

        F.col("product.tags").alias("tags"),

        F.col("product.images").alias("images"),

        F.to_timestamp(
            F.col("product.meta.createdAt")
        ).alias("created_at"),

        F.to_timestamp(
            F.col("product.meta.updatedAt")
        ).alias("updated_at"),

        F.col("product.meta.barcode")
            .alias("barcode"),

        F.col("product.meta.qrCode")
            .alias("qr_code"),

        F.col("product.thumbnail")
            .alias("thumbnail")
    )
    .filter(
        F.col("product_id").isNotNull()
    )
)
display(silver_products)
```

```python
silver_products = (
    silver_products
    .dropDuplicates(["product_id"])
)
display(silver_products)
```
